```r
library(kableExtra)
#library(agricolae)
library(knitr)
#library(faraway)
library(XML)
library(ggplot2)
library(tidyverse)
library(readxl)
# funciones 

## Función que agrega la respuesta a preguntas numéricas 
## toma la respuesa y construye una cadena de caracteres en el formato de 
## respuesta numerica de moodle. ejemplo de uso 
## pregnum(df_r$N,error=0,digresp=0) 
## salida: '{1:NM:=45:0}' esta salida se concatena con el texto de la pregunta 
## y la pregunta, para agregarla al xml   
pregnum=function(resp,error=0.0001,peso=1,digresp=4,diger=digresp){
paste('{',peso,':NM:=',formatC(resp,digits=digresp,format = "f"),':',formatC(error,digits=diger,format = "f"),'}',sep='')
}

## Función que agrega la respuesta a preguntas multichoise 
## toma un vector donde esta el menu de opciones y cual es la respuesta correcta
## a la pregunta y arma el nucleo de la respuesta multichiose. 
## ejemplo de uso 
## opc0=c('is_na(Peso)','mutate','n()','mean(Peso)','parse_number','Peso','summarise','min','Nhermanos','Edad')
## premch(2,opc0) 
## regresa 
## '{1:MULTICHOICE:~%0%is_na(Peso)~=mutate~%0%n()~%0%mean(Peso)~%0%parse_number~%0%Peso~%0%summarise~%0%min~%0%Nhermanos~%0%Edad}'


premch=function(resp=1,opciones,peso=1){
if(resp<1 | resp>length(opciones) ){stop("La respuesta debe ser un numero entre 1 y el numero de opciones")}  
for(i in 1:length(opciones)){
   if(i!=resp){opciones[i]=paste('~%0%',opciones[i],'',sep="")
}else{opciones[i]=paste('~=',opciones[i],'',sep="")} 
    }
gsub(" ", "",paste0( c(paste('{',peso,':MULTICHOICE:'), noquote(opciones),'}' ),collapse=""))
} 
# doc para guardar en formato xml 
NombreCat="Limpieza/parse_num"
doc=newXMLDoc()
root=newXMLNode("quiz",doc=doc)
question=newXMLNode("question",attrs = list(type="category"),parent = root)
category=newXMLNode("category",parent = question)
text=newXMLNode("text",NombreCat,parent = category)
info=newXMLNode("info",attrs = list(format="html"),parent = question)  
textInf=newXMLNode("text","texto",parent = info)
idnum=newXMLNode("idnumber",parent=question)
# funcion para agregar la pregunta al xml 
addquestion=function(NomPr,textopregunta){
questionpr=newXMLNode("question",attrs = list(type="cloze"),parent = root)
nombrepr=newXMLNode("name",parent = questionpr)
textNomPre=newXMLNode("text",NomPr,parent =nombrepr)
questiontext=newXMLNode("questiontext",attrs = list(format="html"),parent = questionpr)
textqt=newXMLNode("text",textopregunta,parent =questiontext)
generalfeedback=newXMLNode("generalfeedback",attrs = list(format="html"),parent = questionpr)
textgf=newXMLNode("text",parent = generalfeedback)
penalty=newXMLNode("penalty",0.3333333,parent=questionpr)
hidden=newXMLNode("hidden",0,parent=questionpr)
idnumber=newXMLNode("idnumber",parent=questionpr)
}


## Leer aer archivo de datos de donde se harán cálculos redactar las preguntas 
## que el estudiante debe responder, mostrando dominio de los temas 
setwd("~/Documents/Cursos/Laboratorio")
# lectura de los datos 
library(readxl)
df=read_excel("~/Documents/Cursos/Laboratorio/DatosEstudiantes.xlsx")
head(df)

# Convertir peso a numérica 
df_limpio=df|>mutate(Peso=parse_number(Peso)) 
#Convertir estatura a numérica
df_limpio=df_limpio|>mutate(Estatura=parse_number(Estatura))
# Estaturas en centímetro, pasarlas a metro (dividir por 100)  
df_limpio <- df_limpio |>
  mutate(Estatura= if_else(
    Estatura > 2.0,
    Estatura / 100,
    Estatura  
  ))
summary(df_limpio$Estatura) 
# Datos faltantes en peso 
Nas=df_limpio|>summarise(NoNa=sum(is.na(Peso))) 
resp0=Nas$NoNa
df_r=df_limpio|>summarise(N=n(),Min=min(Peso,na.rm=TRUE),Q1=quantile(Peso,probs=c(0.25),na.rm=TRUE),Mediana=median(Peso,na.rm=TRUE),Media=mean(Peso,na.rm=TRUE))
df_r

## A continuación se llena una tabla en html con los items que el estudiante debe 
## calcular y responder  


# Peso 
columna='Peso'
## tabla vacía 
tab1='<table style="border-collapse:collapse;border-color:#aaa;border-spacing:0" class="tg"><thead>
<tr><th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">N</th><th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">Min</th><th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">Q1</th>
<th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">Mediana</th><th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">Media</th></tr>
</thead>
<tbody>
<tr><td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp1</td><td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp2</td><td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp3</td><td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp4</td>
<td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp5</td></tr>
</tbody></table>' 

rsp1=pregnum(df_r$N,error=0,digresp=0)
rsp2=pregnum(df_r$Min,error=0,digresp=0)
rsp3=pregnum(df_r$Q1,error=0.01,digresp=2)
rsp4=pregnum(df_r$Mediana,error=0.01,digresp=2)
rsp5=pregnum(df_r$Media,error=0.005,digresp=3)
# se agregan los items a la tabla 
tab1=gsub(pattern = "resp1", replacement = rsp1, x = tab1)
tab1=gsub(pattern = "resp2", replacement = rsp2, x = tab1)
tab1=gsub(pattern = "resp3", replacement = rsp3, x = tab1)
tab1=gsub(pattern = "resp4", replacement = rsp4, x = tab1)
tab1=gsub(pattern = "resp5", replacement = rsp5, x = tab1) 
## En tab1 queda la tabla llena 

## esto es para preguntas multichoise 
opc0=c('is_na(Peso)','mutate','n()','mean(Peso)','parse_number','Peso','summarise','min','Nhermanos','Edad') 
opc1=c('min','mean','median','quantile','max','var','sd','Media','Mediana')  

## a continuación el texto de la pregunta 
## observe como se intercala texto con los resultados de las funciones
## pregnum y premch 
text1p=paste("El archivo <code>DatosEstudiantes.xlsx</code> contiene datos tomados de estudiantes de primer semestre durante un experimento que buscaba evaluar el nivel de satisfacción de los estudiantes con el servicio prestado por la Universidad. Haga la lectura de los datos y responda:")
text2p=paste("Cuántos datos faltantes hay en la columna <code>",columna,"</code>:",pregnum(resp0,digresp=1,error=00),sep=" ")
text3p=paste("Un posible codigo, en lenguaje <code>R</code>, usando la librería <code>tidyverse</code> y el <code>pipe</code> para obtener la respuesta a la pregunta anterior es: <br> <pre> 
library(tidyverse) <br>
library(readxl) <br>
df=read_excel('DatosEstudiantes.xlsx') <br> 
df=",premch(2,opc0),"(",columna,"=",premch(5,opc0),"(",premch(6,opc0),")|>",premch(7,opc0),"(sum(is.na(",columna,")))</pre>") 
text4p=paste( "<br> Complete la siguiente tabla con estadísticas descriptivas básicas de la columna <code>",columna,"</code> <br> " )
text5p=paste("Un posible codigo, en lenguaje <code>R</code>, usando la librería <code>tidyverse</code> y el <code>pipe</code> para obtener la respuesta a la pregunta anterior es: <br> <pre> 
library(tidyverse) <br>
library(readxl) <br> 
df=read_excel('DatosEstudiantes.xlsx') <br>
df|>",premch(7,opc0),"(N=n(),Min=",premch(1,opc1),"(",columna,",na.rm=TRUE),Q1=",premch(4,opc1),"(",columna,",na.rm=TRUE),",premch(9,opc1),"=",premch(3,opc1),"(",columna,",na.rm=TRUE)",premch(8,opc1),"=",premch(2,opc1),"(",columna,",na.rm=TRUE)  </pre>")

## el siguiente es el texto dinal de la pregunta con sus respuestas,
## note como se inserta la tabla  
texto_preg=paste(text1p,text2p,text3p,text4p,tab1,text5p,sep='<br>')

## el nombre que llevará la pregunta 
NomPreg="Filter"
semilla=1
Nombre_pre=paste(NomPreg,semilla,sep=" ") 

### Agrega la pregunta al xml  
addquestion(NomPr=Nombre_pre,textopregunta =texto_preg)


### Otra pregunta 

# No hermanos 
columna='Nhermanos'


Nas=df|>summarise(NoNa=sum(is.na(Nhermanos))) 
resp0=Nas$NoNa
df_r=df|>summarise(N=n(),Min=min(Nhermanos,na.rm=TRUE),Q1=quantile(Nhermanos,probs=c(0.25),na.rm=TRUE),Mediana=median(Nhermanos,na.rm=TRUE),Media=mean(Nhermanos,na.rm=TRUE))
df_r

tab1='<table style="border-collapse:collapse;border-color:#aaa;border-spacing:0" class="tg"><thead>
<tr><th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">N</th><th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">Min</th><th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">Q1</th>
<th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">Mediana</th><th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">Media</th></tr>
</thead>
<tbody>
<tr><td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp1</td><td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp2</td><td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp3</td><td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp4</td>
<td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;overflow:hidden;padding:10px 5px;text-align:left;vertical-align:top;word-break:normal">resp5</td></tr>
</tbody></table>' 

rsp1=pregnum(df_r$N,error=0,digresp=0)
rsp1
rsp2=pregnum(df_r$Min,error=0,digresp=0)
rsp2
rsp3=pregnum(df_r$Q1,error=0.01,digresp=2)
rsp3
rsp4=pregnum(df_r$Mediana,error=0.01,digresp=2)
rsp4
rsp5=pregnum(df_r$Media,error=0.005,digresp=3)
rsp5
tab1=gsub(pattern = "resp1", replacement = rsp1, x = tab1)
tab1=gsub(pattern = "resp2", replacement = rsp2, x = tab1)
tab1=gsub(pattern = "resp3", replacement = rsp3, x = tab1)
tab1=gsub(pattern = "resp4", replacement = rsp4, x = tab1)
tab1=gsub(pattern = "resp5", replacement = rsp5, x = tab1) 

text1p=paste("El archivo <code>DatosEstudiantes.xlsx</code> contiene datos tomados de estudiantes de primer semestre durante un experimento que buscaba evaluar el nivel de satisfacción de los estudiantes con el servicio prestado por la Universidad. Haga la lectura de los datos y responda:")
text2p=paste("Cuántos datos faltantes hay en la columna <code>",columna,"</code>:",pregnum(resp0,digresp=1,error=00),sep=" ")
text3p=paste("Un posible codigo, en lenguaje <code>R</code>, usando la librería <code>tidyverse</code> y el <code>pipe</code> para obtener la respuesta a la pregunta anterior es: <br> <pre> 
library(tidyverse) <br>
library(readxl) <br>
df=read_excel('DatosEstudiantes.xlsx') <br> 
df=",premch(2,opc0),"(",columna,"=",premch(5,opc0),"(",premch(9,opc0),")|>",premch(7,opc0),"(sum(is.na(",columna,")))</pre>") 
text4p=paste( "<br> Complete la siguiente tabla con estadísticas descriptivas básicas de la columna <code>",columna,"</code> <br> " )
text5p=paste("Un posible codigo, en lenguaje <code>R</code>, usando la librería <code>tidyverse</code> y el <code>pipe</code> para obtener la respuesta a la pregunta anterior es: <br> <pre> 
library(tidyverse) <br>
library(readxl) <br> 
df=read_excel('DatosEstudiantes.xlsx') <br>
df|>",premch(7,opc0),"(N=n(),Min=",premch(1,opc1),"(",columna,",na.rm=TRUE),Q1=",premch(4,opc1),"(",columna,",na.rm=TRUE),",premch(9,opc1),"=",premch(3,opc1),"(",columna,",na.rm=TRUE)",premch(8,opc1),"=",premch(2,opc1),"(",columna,",na.rm=TRUE)  </pre>")

texto_preg=paste(text1p,text2p,text3p,text4p,tab1,text5p,sep='<br>')
NomPreg="Filter"
semilla=2
Nombre_pre=paste(NomPreg,semilla,sep=" ") 

# agrega pregunta 
addquestion(NomPr=Nombre_pre,textopregunta =texto_preg)

##### se agregan muchas mas preguntas de ese estilo 

## se guiarda el xml 
saveXML(doc, file="PreguntasParse.xml") 
```
