```r
library(XML)
library(tidyverse)
library(glue)

# ==============================================================================
# 1. GENERADORES ATÓMICOS DE SINTAXIS CLOZE (Con pesos enteros garantizados)
# ==============================================================================

cloze_num <- function(resp, error = 0.0001, peso = 1, digresp = 4, diger = digresp) {
  peso_int <- as.integer(peso)
  val_formated <- formatC(as.numeric(resp), digits = digresp, format = "f")
  err_formated <- formatC(as.numeric(error), digits = diger, format = "f")
  as.character(glue("{{{peso_int}:NM:={val_formated}:{err_formated}}}"))
}

cloze_mch <- function(resp = 1, opciones, peso = 1, tipo = "MULTICHOICE") {
  if (resp < 1 | resp > length(opciones)) stop("Índice de respuesta fuera de rango.")
  peso_int <- as.integer(peso)
  opciones_formateadas <- map2_chr(opciones, seq_along(opciones), function(opc, idx) {
    if (idx != resp) paste0("~%0%", opc) else paste0("~=", opc)
  })
  as.character(glue("{{{peso_int}:{tipo}:{paste0(opciones_formateadas, collapse = '')}}}"))
}

cloze_sa <- function(respuestas_correctas, peso = 1, case_sensitive = TRUE) {
  # Genera pregunta de respuesta corta Cloze: {peso_entero:SAC:=resp1~=resp2}
  peso_int <- as.integer(peso)
  tipo <- if (case_sensitive) "SAC" else "SA"
  resp_str <- paste0("~=", respuestas_correctas, collapse = "")
  resp_str <- sub("^~", "", resp_str) # Quita el primer '~'
  as.character(glue("{{{peso_int}:{tipo}:{resp_str}}}"))
}

# ==============================================================================
# 2. INICIALIZADOR DEL CONTENEDOR XML Y FUNCIÓN DE GUARDADO
# ==============================================================================

iniciar_quiz_xml <- function(nombre_categoria) {
  doc <- newXMLDoc()
  root <- newXMLNode("quiz", doc = doc)
  
  question_cat <- newXMLNode("question", attrs = list(type = "category"), parent = root)
  category <- newXMLNode("category", parent = question_cat)
  newXMLNode("text", nombre_categoria, parent = category)
  info <- newXMLNode("info", attrs = list(format = "html"), parent = question_cat)  
  newXMLNode("text", "Banco de preguntas generado automáticamente desde R", parent = info)
  newXMLNode("idnumber", parent = question_cat)
  
  return(list(doc = doc, root = root))
}

crear_pregunta_cloze <- function(moodle_quiz, nombre_pregunta, plantilla_texto, lista_respuestas, plantilla_tabla = NULL) {
  if (!is.null(plantilla_tabla)) {
    tabla_renderizada <- glue_data(lista_respuestas, plantilla_tabla, .open = "[[", .close = "]]")
    lista_respuestas$TABLA_EVALUACION <- as.character(tabla_renderizada)
  }
  
  texto_final_html <- glue_data(lista_respuestas, plantilla_texto, .open = "[[", .close = "]]")
  
  questionpr <- newXMLNode("question", attrs = list(type = "cloze"), parent = moodle_quiz$root)
  nombrepr <- newXMLNode("name", parent = questionpr)
  newXMLNode("text", nombre_pregunta, parent = nombrepr)
  
  questiontext <- newXMLNode("questiontext", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", as.character(texto_final_html), parent = questiontext)
  
  generalfeedback <- newXMLNode("generalfeedback", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", "", parent = generalfeedback)
  
  newXMLNode("penalty", "0.3333333", parent = questionpr)
  newXMLNode("hidden", "0", parent = questionpr)
  newXMLNode("idnumber", parent = questionpr)
}

guardar_quiz_xml <- function(moodle_quiz, archivo_salida) {
  saveXML(moodle_quiz$doc, file = archivo_salida, encoding = "UTF-8")
  cat(glue("Archivo '{archivo_salida}' generado exitosamente.\n"))
}

# ==============================================================================
# 3. GENERADOR DE TABLA HTML PARA DOS MUESTRAS DE DISTINTO TAMAÑO (n1 y n2)
# ==============================================================================

generar_tabla_muestras_html <- function(x1, x2) {
  max_n <- max(length(x1), length(x2))
  # Rellenamos con celdas vacías si una muestra es más corta que la otra
  x1_str <- c(formatC(x1, digits = 2, format = "f"), rep("", max_n - length(x1)))
  x2_str <- c(formatC(x2, digits = 2, format = "f"), rep("", max_n - length(x2)))
  
  filas <- map_chr(seq_len(max_n), function(idx) {
    glue("  <tr>
    <td class='tg-celda' style='background-color:#f4f6f9; font-weight:bold;'>{idx}</td>
    <td class='tg-celda'>{x1_str[idx]}</td>
    <td class='tg-celda'>{x2_str[idx]}</td>
  </tr>")
  })
  
  glue('<style type="text/css">
.tg-datos {border-collapse:collapse; border-color:#9ABAD9; border-spacing:0; margin: 12px 0;}
.tg-datos th {background-color:#409cff; border:1px solid #9ABAD9; color:#fff; font-family:Arial, sans-serif; font-size:14px; padding:8px 15px; text-align:center;}
.tg-datos .tg-celda {background-color:#EBF5FF; border:1px solid #9ABAD9; color:#333; font-family:monospace; font-size:14px; padding:6px 15px; text-align:center;}
</style>
<table class="tg-datos">
<thead>
  <tr>
    <th>Observación</th>
    <th>Plástico 1 (psi)</th>
    <th>Plástico 2 (psi)</th>
  </tr>
</thead>
<tbody>
[[paste0(filas, collapse = "\n")]]
</tbody>
</table>', .open = "[[", .close = "]]")
}

# ==============================================================================
# 4. PLANTILLA DEL ENUNCIADO DE PRUEBA DE HIPÓTESIS EN R
# ==============================================================================

plantilla_enunciado_ph <- '
<p>Existen dos tipos de plástico para su uso por un fabricante de componentes electrónicos. La tensión de ruptura de este plástico es un parámetro importante. Se sabe que las desviaciones estándar poblacionales son \\(\\sigma_1 = [[sigma1]]\\) psi y \\(\\sigma_2 = [[sigma2]]\\) psi.</p>

<p>A continuación se registran en la tabla los datos de tensión de ruptura (en psi) obtenidos de dos muestras aleatorias independientes (para copiarlos rápidamente a R, también puede usar los vectores del bloque de código):</p>

[[TABLA_DATOS]]

<pre style="background-color:#f4f6f9; border-left:4px solid #409cff; padding:10px; font-family:monospace; font-size:13px; color:#222;"><code>plastico1 &lt;- c([[vec_x1]])
plastico2 &lt;- c([[vec_x2]])</code></pre>

<p>La compañía <b>no adoptará el Plástico 1 a menos que</b> la tensión de ruptura promedio de este exceda a la del Plástico 2 al menos por <b>[[delta0]]</b> psi. Con base en la información contenida en la muestra y usando un nivel de significancia \\(\\alpha = [[alfa]]\\), utilice <b>R</b> para responder:</p>

<ol>
  <li><b>Formulación de hipótesis:</b> Para demostrar estadísticamente que el Plástico 1 cumple la exigencia de la compañía, contrastamos:<br>
      \\(H_0: \\mu_1 - \\mu_2 \\le [[delta0]]\\) &nbsp; frente a &nbsp; \\(H_1: \\mu_1 - \\mu_2\\) [[cloze_h1]] <b>[[delta0]]</b>.</li>

  <li><b>Estadísticos muestrales en R:</b> Usando las funciones <code>length()</code> y <code>mean()</code> en R, determine los tamaños de muestra y las medias muestrales:<br>
      <ul>
        <li>Plástico 1: \\(n_1 = \\) [[cloze_n1]] &nbsp; y &nbsp; \\(\\bar{x}_1 = \\) [[cloze_media1]] psi (a 2 decimales).</li>
        <li>Plástico 2: \\(n_2 = \\) [[cloze_n2]] &nbsp; y &nbsp; \\(\\bar{x}_2 = \\) [[cloze_media2]] psi (a 2 decimales).</li>
      </ul>
  </li>

  <li><b>Estadístico de prueba y valor Crítico en R:</b><br>
      <ul>
        <li>El error estándar de la diferencia de medias \\(\\sqrt{\\frac{\\sigma_1^2}{n_1} + \\frac{\\sigma_2^2}{n_2}}\\) es: [[cloze_se]] (a 4 decimales).</li>
        <li>El valor del estadístico de prueba \\(Z_0 = \\frac{(\\bar{x}_1 - \\bar{x}_2) - [[delta0]]}{EE}\\) es: [[cloze_z0]] (a 4 decimales).</li>
        <li>¿Qué función de R base devuelve el cuantil (valor crítico \\(z_{1-\\alpha}\\)) de la distribución normal estándar?: <code>[[cloze_func_q]]([[uno_menos_alfa]])</code>, cuyo valor numérico es [[cloze_z_crit]] (a 4 decimales).</li>
      </ul>
  </li>

  <li><b>Valor-p en R y Decisión:</b><br>
      <ul>
        <li>¿Qué función de R base permite calcular la probabilidad de cola superior \\(P(Z > Z_0)\\) usando <code>lower.tail = FALSE</code>?: <code>[[cloze_func_p]](Z0, lower.tail = FALSE)</code>.</li>
        <li>El valor-p obtenido para esta prueba es: [[cloze_pval]] (a 4 decimales).</li>
        <li>Con un nivel de significancia \\(\\alpha = [[alfa]]\\), se [[cloze_rechazo]] la hipótesis nula \\(H_0\\). Por lo tanto, la compañía [[cloze_adopcion]] adoptar el Plástico 1.</li>
      </ul>
  </li>
</ol>
'

# ==============================================================================
# 5. BUCLE GENERADOR DEL BANCO DE 15 PREGUNTAS EN R
# ==============================================================================

banco_moodle <- iniciar_quiz_xml(nombre_categoria = "CursoR/Pruebas_Hipotesis_Dos_Muestras_Z")

opc_signo    <- c(">", "<", "\\neq", "=")
opc_rechazo  <- c("rechaza", "no rechaza")
opc_adopcion <- c("sí deberá", "no deberá")

semillas <- seq(501, by = 13, length.out = 15)

for (i in seq_along(semillas)) {
  sem <- semillas[i]
  set.seed(sem)
  
  # 1. Parámetros variables del problema
  n1     <- sample(8:14, 1)
  n2     <- sample(8:14, 1)
  sigma1 <- sample(c(1.0, 1.2, 1.5, 1.8, 2.0), 1)
  sigma2 <- sample(c(1.0, 1.2, 1.5, 1.8, 2.0), 1)
  delta0 <- sample(c(8, 10, 12, 15), 1)
  alfa   <- sample(c(0.01, 0.05, 0.10), 1)
  
  # 2. Alternamos entre casos donde H0 se rechaza (i impar) y donde NO se rechaza (i par)
  # Así evitamos que los estudiantes adivinen siempre la misma conclusión final
  se_teorico <- sqrt((sigma1^2 / n1) + (sigma2^2 / n2))
  z_objetivo <- if (i %% 2 != 0) runif(1, 1.95, 3.10) else runif(1, -1.80, 1.15)
  
  mu2_sim <- sample(150:165, 1)
  mu1_sim <- mu2_sim + delta0 + z_objetivo * se_teorico
  
  # 3. Simulamos las muestras y redondeamos a 2 decimales (tal como se verán en la tabla)
  x1 <- round(rnorm(n1, mean = mu1_sim, sd = sigma1), 2)
  x2 <- round(rnorm(n2, mean = mu2_sim, sd = sigma2), 2)
  
  # 4. Cálculos exactos sobre los vectores redondeados (los mismos que obtendrá el alumno)
  media1 <- mean(x1)
  media2 <- mean(x2)
  se_dif <- sqrt((sigma1^2 / n1) + (sigma2^2 / n2))
  z0     <- ((media1 - media2) - delta0) / se_dif
  z_crit <- qnorm(1 - alfa)
  pval   <- pnorm(z0, lower.tail = FALSE)
  
  # Índice correcto de decisión (1 = rechaza / sí deberá; 2 = no rechaza / no deberá)
  idx_decision <- if (pval < alfa) 1 else 2
  
  # 5. Construimos la lista de respuestas e insumos para la plantilla
  respuestas_iter <- list(
    sigma1         = formatC(sigma1, digits = 1, format = "f"),
    sigma2         = formatC(sigma2, digits = 1, format = "f"),
    delta0         = delta0,
    alfa           = alfa,
    uno_menos_alfa = 1 - alfa,
    TABLA_DATOS    = generar_tabla_muestras_html(x1, x2),
    vec_x1         = paste(formatC(x1, digits = 2, format = "f"), collapse = ", "),
    vec_x2         = paste(formatC(x2, digits = 2, format = "f"), collapse = ", "),
    
    # Campos Cloze
    cloze_h1       = cloze_mch(resp = 1, opciones = opc_signo, peso = 1),
    cloze_n1       = cloze_num(n1, digresp = 0, error = 0, peso = 1),
    cloze_n2       = cloze_num(n2, digresp = 0, error = 0, peso = 1),
    cloze_media1   = cloze_num(media1, digresp = 2, error = 0.02, peso = 2),
    cloze_media2   = cloze_num(media2, digresp = 2, error = 0.02, peso = 2),
    cloze_se       = cloze_num(se_dif, digresp = 4, error = 0.002, peso = 2),
    cloze_z0       = cloze_num(z0, digresp = 4, error = 0.02, peso = 2),
    cloze_func_q   = cloze_sa(c("qnorm", "stats::qnorm"), peso = 1),
    cloze_z_crit   = cloze_num(z_crit, digresp = 4, error = 0.005, peso = 1),
    cloze_func_p   = cloze_sa(c("pnorm", "stats::pnorm"), peso = 1),
    cloze_pval     = cloze_num(pval, digresp = 4, error = 0.005, peso = 2),
    cloze_rechazo  = cloze_mch(resp = idx_decision, opciones = opc_rechazo, peso = 1),
    cloze_adopcion = cloze_mch(resp = idx_decision, opciones = opc_adopcion, peso = 1)
  )
  
  nombre_preg <- glue("PH_DosMuestrasZ_Plastico_v{sprintf('%02d', i)}_n{n1}_{n2}")
  
  crear_pregunta_cloze(
    moodle_quiz      = banco_moodle,
    nombre_pregunta  = nombre_preg,
    plantilla_texto  = plantilla_enunciado_ph,
    lista_respuestas = respuestas_iter
  )
}

# 6. Exportamos el banco XML final listo para Moodle
guardar_quiz_xml(banco_moodle, "Banco_15_Preguntas_Hipotesis_Plasticos_R.xml")
```

```r
library(XML)
library(tidyverse)
library(glue)

# ==============================================================================
# 1. GENERADORES ATÓMICOS DE SINTAXIS CLOZE (Con pesos enteros garantizados)
# ==============================================================================

cloze_num <- function(resp, error = 0.0001, peso = 1, digresp = 4, diger = digresp) {
  peso_int <- as.integer(peso)
  val_formated <- formatC(as.numeric(resp), digits = digresp, format = "f")
  err_formated <- formatC(as.numeric(error), digits = diger, format = "f")
  as.character(glue("{{{peso_int}:NM:={val_formated}:{err_formated}}}"))
}

cloze_mch <- function(resp = 1, opciones, peso = 1, tipo = "MULTICHOICE") {
  if (resp < 1 | resp > length(opciones)) stop("Índice de respuesta fuera de rango.")
  peso_int <- as.integer(peso)
  opciones_formateadas <- map2_chr(opciones, seq_along(opciones), function(opc, idx) {
    if (idx != resp) paste0("~%0%", opc) else paste0("~=", opc)
  })
  as.character(glue("{{{peso_int}:{tipo}:{paste0(opciones_formateadas, collapse = '')}}}"))
}

cloze_sa <- function(respuestas_correctas, peso = 1, case_sensitive = TRUE) {
  peso_int <- as.integer(peso)
  tipo <- if (case_sensitive) "SAC" else "SA"
  resp_str <- paste0("~=", respuestas_correctas, collapse = "")
  resp_str <- sub("^~", "", resp_str)
  as.character(glue("{{{peso_int}:{tipo}:{resp_str}}}"))
}

# ==============================================================================
# 2. INICIALIZADOR DEL CONTENEDOR XML Y GUARDADO
# ==============================================================================

iniciar_quiz_xml <- function(nombre_categoria) {
  doc <- newXMLDoc()
  root <- newXMLNode("quiz", doc = doc)
  
  question_cat <- newXMLNode("question", attrs = list(type = "category"), parent = root)
  category <- newXMLNode("category", parent = question_cat)
  newXMLNode("text", nombre_categoria, parent = category)
  info <- newXMLNode("info", attrs = list(format = "html"), parent = question_cat)  
  newXMLNode("text", "Banco de preguntas Prueba T generado automáticamente desde R", parent = info)
  newXMLNode("idnumber", parent = question_cat)
  
  return(list(doc = doc, root = root))
}

crear_pregunta_cloze <- function(moodle_quiz, nombre_pregunta, plantilla_texto, lista_respuestas, plantilla_tabla = NULL) {
  if (!is.null(plantilla_tabla)) {
    tabla_renderizada <- glue_data(lista_respuestas, plantilla_tabla, .open = "[[", .close = "]]")
    lista_respuestas$TABLA_EVALUACION <- as.character(tabla_renderizada)
  }
  
  texto_final_html <- glue_data(lista_respuestas, plantilla_texto, .open = "[[", .close = "]]")
  
  questionpr <- newXMLNode("question", attrs = list(type = "cloze"), parent = moodle_quiz$root)
  nombrepr <- newXMLNode("name", parent = questionpr)
  newXMLNode("text", nombre_pregunta, parent = nombrepr)
  
  questiontext <- newXMLNode("questiontext", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", as.character(texto_final_html), parent = questiontext)
  
  generalfeedback <- newXMLNode("generalfeedback", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", "", parent = generalfeedback)
  
  newXMLNode("penalty", "0.3333333", parent = questionpr)
  newXMLNode("hidden", "0", parent = questionpr)
  newXMLNode("idnumber", parent = questionpr)
}

guardar_quiz_xml <- function(moodle_quiz, archivo_salida) {
  saveXML(moodle_quiz$doc, file = archivo_salida, encoding = "UTF-8")
  cat(glue("Archivo '{archivo_salida}' generado exitosamente.\n"))
}

# ==============================================================================
# 3. GENERADOR DE TABLA HTML PARA DOS MUESTRAS (Temperatura de Deflexión)
# ==============================================================================

generar_tabla_temperaturas_html <- function(x1, x2) {
  max_n <- max(length(x1), length(x2))
  x1_str <- c(formatC(x1, digits = 1, format = "f"), rep("", max_n - length(x1)))
  x2_str <- c(formatC(x2, digits = 1, format = "f"), rep("", max_n - length(x2)))
  
  filas <- map_chr(seq_len(max_n), function(idx) {
    glue("  <tr>
    <td class='tg-celda' style='background-color:#f4f6f9; font-weight:bold;'>{idx}</td>
    <td class='tg-celda'>{x1_str[idx]}</td>
    <td class='tg-celda'>{x2_str[idx]}</td>
  </tr>")
  })
  
  glue('<style type="text/css">
.tg-datos {border-collapse:collapse; border-color:#9ABAD9; border-spacing:0; margin: 12px 0;}
.tg-datos th {background-color:#2e7d32; border:1px solid #9ABAD9; color:#fff; font-family:Arial, sans-serif; font-size:14px; padding:8px 15px; text-align:center;}
.tg-datos .tg-celda {background-color:#e8f5e9; border:1px solid #9ABAD9; color:#333; font-family:monospace; font-size:14px; padding:6px 15px; text-align:center;}
</style>
<table class="tg-datos">
<thead>
  <tr>
    <th>Espécimen</th>
    <th>Solución 1 (°C)</th>
    <th>Solución 2 (°C)</th>
  </tr>
</thead>
<tbody>
[[paste0(filas, collapse = "\n")]]
</tbody>
</table>', .open = "[[", .close = "]]")
}

# ==============================================================================
# 4. PLANTILLA DEL ENUNCIADO (Prueba T con varianzas desconocidas iguales)
# ==============================================================================

plantilla_enunciado_t <- '
<p>Se investigan las temperaturas de deflexión bajo carga (en °C) para dos tipos de tubería de plástico (Solución 1 y Solución 2). Para ello se toman dos muestras aleatorias independientes, anotando las temperaturas observadas:</p>

[[TABLA_DATOS]]

<pre style="background-color:#f4f6f9; border-left:4px solid #2e7d32; padding:10px; font-family:monospace; font-size:13px; color:#222;"><code>solucion1 &lt;- c([[vec_x1]])
solucion2 &lt;- c([[vec_x2]])</code></pre>

<p>Asuma que las poblaciones se distribuyen de manera normal y que las varianzas son desconocidas pero iguales \\((\\sigma_1^2 = \\sigma_2^2)\\). Se desea contrastar la hipótesis nula \\(H_0: \\mu_1 - \\mu_2 = [[delta0]]\\) frente a la alternativa \\(H_1: \\mu_1 - \\mu_2 \\neq [[delta0]]\\) usando un nivel de significancia \\(\\alpha = [[alfa]]\\).</p>

<ol>
  <li><b>Estimador de la varianza común:</b> Usando las muestras, calcule un estimador insesgado de la varianza común \\(s_p^2\\): [[cloze_sp2]] (a 4 decimales).</li>

  <li><b>Estadístico de prueba:</b> El valor del estadístico \\(t_0\\) para esta prueba es: [[cloze_t0]] (a 3 decimales).</li>

  <li><b>Valor Crítico:</b> El valor crítico de la distribución \\(t\\) de Student con \\(df = [[df]]\\) grados de libertad (para una prueba bilateral al nivel \\(\\alpha = [[alfa]]\\)) es: [[cloze_tcrit]] (a 3 decimales).</li>

  <li><b>Decisión estadística:</b> Comparando el valor absoluto del estadístico con el valor crítico, se concluye que: [[cloze_decision]]</li>
</ol>
'

# ==============================================================================
# 5. BUCLE GENERADOR DEL BANCO DE 15 PREGUNTAS EN R (Prueba T)
# ==============================================================================

banco_moodle <- iniciar_quiz_xml(nombre_categoria = "CursoR/Pruebas_Hipotesis_Dos_Muestras_T")

opciones_decision <- c(
  "se rechaza la hipótesis nula (existe diferencia significativa)",
  "no se rechaza la hipótesis nula (no hay evidencia suficiente de diferencia)"
)

semillas_t <- seq(701, by = 17, length.out = 15)

for (i in seq_along(semillas_t)) {
  sem <- semillas_t[i]
  set.seed(sem)
  
  # Parámetros aleatorios de la simulación
  n1     <- sample(8:12, 1)
  n2     <- sample(8:12, 1)
  sd_comun <- sample(c(1.2, 1.5, 1.8, 2.1, 2.5), 1)
  delta0 <- sample(c(0, 2, 4), 1)
  alfa   <- sample(c(0.01, 0.05), 1)
  
  # Forzamos alternancia entre rechazar (i impar) y no rechazar (i par)
  t_objetivo <- if (i %% 2 != 0) runif(1, 2.2, 3.8) else runif(1, 0.2, 1.5)
  
  mu2 <- sample(190:198, 1)
  se_teorico <- sd_comun * sqrt(1/n1 + 1/n2)
  mu1 <- mu2 + delta0 + t_objetivo * se_teorico
  
  # Simulamos y redondeamos a 1 decimal
  x1 <- round(rnorm(n1, mean = mu1, sd = sd_comun), 1)
  x2 <- round(rnorm(n2, mean = mu2, sd = sd_comun), 1)
  
  # --- Cálculos exactos basados en los datos mostrados ---
  m1 <- mean(x1)
  m2 <- mean(x2)
  v1 <- var(x1)
  v2 <- var(x2)
  df <- n1 + n2 - 2
  
  # Varianza común (pooled variance)
  s_p2 <- ((n1 - 1) * v1 + (n2 - 1) * v2) / df
  
  # Estadístico t0
  t0 <- ((m1 - m2) - delta0) / sqrt(s_p2 * (1/n1 + 1/n2))
  
  # Valor crítico bilateral
  t_crit <- qt(1 - alfa / 2, df)
  
  # Decisión (1 = Rechaza H0, 2 = No rechaza H0)
  idx_dec <- if (abs(t0) > t_crit) 1 else 2
  
  respuestas_iter <- list(
    delta0       = delta0,
    alfa         = alfa,
    df           = df,
    TABLA_DATOS  = generar_tabla_temperaturas_html(x1, x2),
    vec_x1       = paste(formatC(x1, digits = 1, format = "f"), collapse = ", "),
    vec_x2       = paste(formatC(x2, digits = 1, format = "f"), collapse = ", "),
    
    # Campos Cloze
    cloze_sp2    = cloze_num(s_p2, digresp = 4, error = 0.05, peso = 2),
    cloze_t0     = cloze_num(t0, digresp = 3, error = 0.05, peso = 2),
    cloze_tcrit  = cloze_num(t_crit, digresp = 3, error = 0.05, peso = 1),
    cloze_decision = cloze_mch(resp = idx_dec, opciones = opciones_decision, peso = 1)
  )
  
  nombre_preg <- glue("PH_DosMuestrasT_Deflexion_v{sprintf('%02d', i)}_df{df}")
  
  crear_pregunta_cloze(
    moodle_quiz      = banco_moodle,
    nombre_pregunta  = nombre_preg,
    plantilla_texto  = plantilla_enunciado_t,
    lista_respuestas = respuestas_iter
  )
}

# 6. Exportamos el banco XML de Prueba T
guardar_quiz_xml(banco_moodle, "Banco_15_Preguntas_Hipotesis_T_Student_R.xml")
```

## Prueba T pareada 

```r
library(XML)
library(tidyverse)
library(glue)

# ==============================================================================
# 1. GENERADORES ATÓMICOS DE SINTAXIS CLOZE (Pesos enteros garantizados)
# ==============================================================================

cloze_num <- function(resp, error = 0.0001, peso = 1, digresp = 4, diger = digresp) {
  peso_int <- as.integer(peso)
  val_formated <- formatC(as.numeric(resp), digits = digresp, format = "f")
  err_formated <- f_err <- formatC(as.numeric(error), digits = diger, format = "f")
  as.character(glue("{{{peso_int}:NM:={val_formated}:{err_formated}}}"))
}

cloze_mch <- function(resp = 1, opciones, peso = 1, tipo = "MULTICHOICE") {
  if (resp < 1 | resp > length(opciones)) stop("Índice de respuesta fuera de rango.")
  peso_int <- as.integer(peso)
  opciones_formateadas <- map2_chr(opciones, seq_along(opciones), function(opc, idx) {
    if (idx != resp) paste0("~%0%", opc) else paste0("~=", opc)
  })
  as.character(glue("{{{peso_int}:{tipo}:{paste0(opciones_formateadas, collapse = '')}}}"))
}

cloze_sa <- function(respuestas_correctas, peso = 1, case_sensitive = TRUE) {
  peso_int <- as.integer(peso)
  tipo <- if (case_sensitive) "SAC" else "SA"
  resp_str <- paste0("~=", respuestas_correctas, collapse = "")
  resp_str <- sub("^~", "", resp_str)
  as.character(glue("{{{peso_int}:{tipo}:{resp_str}}}"))
}

# ==============================================================================
# 2. INICIALIZADOR DEL CONTENEDOR XML Y GUARDADO
# ==============================================================================

iniciar_quiz_xml <- function(nombre_categoria) {
  doc <- newXMLDoc()
  root <- newXMLNode("quiz", doc = doc)
  
  question_cat <- newXMLNode("question", attrs = list(type = "category"), parent = root)
  category <- newXMLNode("category", parent = question_cat)
  newXMLNode("text", nombre_categoria, parent = category)
  info <- newXMLNode("info", attrs = list(format = "html"), parent = question_cat)  
  newXMLNode("text", "Banco de preguntas Prueba T Pareada generado automáticamente desde R", parent = info)
  newXMLNode("idnumber", parent = question_cat)
  
  return(list(doc = doc, root = root))
}

crear_pregunta_cloze <- function(moodle_quiz, nombre_pregunta, plantilla_texto, lista_respuestas, plantilla_tabla = NULL) {
  if (!is.null(plantilla_tabla)) {
    tabla_renderizada <- glue_data(lista_respuestas, plantilla_tabla, .open = "[[", .close = "]]")
    lista_respuestas$TABLA_EVALUACION <- as.character(tabla_renderizada)
  }
  
  texto_final_html <- glue_data(lista_respuestas, plantilla_texto, .open = "[[", .close = "]]")
  
  questionpr <- newXMLNode("question", attrs = list(type = "cloze"), parent = moodle_quiz$root)
  nombrepr <- newXMLNode("name", parent = questionpr)
  newXMLNode("text", nombre_pregunta, parent = nombrepr)
  
  questiontext <- newXMLNode("questiontext", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", as.character(texto_final_html), parent = questiontext)
  
  generalfeedback <- newXMLNode("generalfeedback", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", "", parent = generalfeedback)
  
  newXMLNode("penalty", "0.3333333", parent = questionpr)
  newXMLNode("hidden", "0", parent = questionpr)
  newXMLNode("idnumber", parent = questionpr)
}

guardar_quiz_xml <- function(moodle_quiz, archivo_salida) {
  saveXML(moodle_quiz$doc, file = archivo_salida, encoding = "UTF-8")
  cat(glue("Archivo '{archivo_salida}' generado exitosamente.\n"))
}

# ==============================================================================
# 3. GENERADOR DE TABLA HTML HORIZONTAL COMPACTA (Sujetos 1 a n)
# ==============================================================================
# Como son entre 25 y 38 sujetos, presentamos la tabla en formato horizontal 
# (o dividida en bloques de 12 columnas) para que ocupe poco espacio vertical en Moodle.

generar_tabla_pareada_html <- function(md, mi) {
  n <- length(md)
  tam_bloque <- 12
  num_bloques <- ceiling(n / tam_bloque)
  
  tablas_bloques <- map_chr(seq_len(num_bloques), function(b) {
    idx_ini <- (b - 1) * tam_bloque + 1
    idx_fin <- min(b * tam_bloque, n)
    indices <- idx_ini:idx_fin
    
    ths_sujeto <- paste0("<th>", indices, "</th>", collapse = "")
    tds_md     <- paste0("<td class='tg-celda'>", md[indices], "</td>", collapse = "")
    tds_mi     <- paste0("<td class='tg-celda'>", mi[indices], "</td>", collapse = "")
    
    glue("<table class='tg-datos' style='margin-bottom:8px;'>
  <thead>
    <tr><th style='background-color:#1565c0;'>Sujeto</th>{ths_sujeto}</tr>
  </thead>
  <tbody>
    <tr><td class='tg-etq'>Mano Derecha (<code>fuerza_md</code>)</td>{tds_md}</tr>
    <tr><td class='tg-etq'>Mano Izquierda (<code>fuerza_mi</code>)</td>{tds_mi}</tr>
  </tbody>
</table>")
  })
  
  estilo <- '<style type="text/css">
.tg-datos {border-collapse:collapse; border-color:#9ABAD9; border-spacing:0;}
.tg-datos th {background-color:#1976d2; border:1px solid #9ABAD9; color:#fff; font-family:Arial, sans-serif; font-size:13px; padding:5px 8px; text-align:center;}
.tg-datos .tg-etq {background-color:#f4f6f9; border:1px solid #9ABAD9; color:#222; font-family:Arial, sans-serif; font-size:13px; font-weight:bold; padding:5px 10px; text-align:left;}
.tg-datos .tg-celda {background-color:#EBF5FF; border:1px solid #9ABAD9; color:#333; font-family:monospace; font-size:14px; padding:5px 8px; text-align:center;}
</style>\n'
  
  paste0(estilo, paste0(tablas_bloques, collapse = "\n"))
}

# ==============================================================================
# 4. PLANTILLA DEL ENUNCIADO (Prueba T Pareada en R)
# ==============================================================================

plantilla_enunciado_pareada <- '
<p>En un estudio de fisiología del ejercicio se midió la fuerza prensil (en kg) tanto de la mano derecha (<code>fuerza_md</code>) como de la mano izquierda (<code>fuerza_mi</code>) en una muestra aleatoria de <b>[[n]]</b> individuos. Los datos obtenidos para cada sujeto se presentan a continuación:</p>

[[TABLA_DATOS]]

<p>Para cargar rápidamente estos datos en su sesión de <b>R</b>, puede copiar y ejecutar el siguiente bloque de código:</p>

<pre style="background-color:#f4f6f9; border-left:4px solid #1976d2; padding:10px; font-family:monospace; font-size:13px; color:#222; overflow-x:auto;"><code>fuerza_md &lt;- c([[vec_md]])
fuerza_mi &lt;- c([[vec_mi]])</code></pre>

<div style="background-color:#fff8e1; border-left:4px solid #ffa000; padding:10px; margin:12px 0;">
  <b>Pregunta de investigación:</b> Con base en la evidencia muestral y usando un nivel de significancia \\(\\alpha = [[alfa]]\\), [[pregunta_investigacion]]<br>
  <em>(Defina las diferencias como \\(D = \\text{Mano Derecha} - \\text{Mano Izquierda}\\), con media poblacional \\(\\mu_D = \\mu_{\\text{MD}} - \\mu_{\\text{MI}}\\)).</em>
</div>

<ol>
  <li><b>Formulación de hipótesis y código en R:</b><br>
      <ul>
        <li>Para responder a la pregunta de investigación, se contrasta la hipótesis nula \\(H_0: \\mu_D = 0\\) frente a la hipótesis alternativa \\(H_1: \\mu_D\\) [[cloze_signo_h1]] \\(0\\).</li>
        <li>Al ejecutar la función <code>t.test(fuerza_md, fuerza_mi, paired = [[cloze_bool_paired]], alternative = "[[cloze_arg_alt]]")</code> en R, el promedio de las diferencias muestrales \\(\\bar{d}\\) es: [[cloze_d_bar]] kg (a 3 decimales).</li>
      </ul>
  </li>

  <li><b>Estadístico de prueba y grados de libertad:</b><br>
      <ul>
        <li>El valor del estadístico de prueba \\(t_0\\) es: [[cloze_t0]] (a 3 decimales).</li>
        <li>Los grados de libertad \\(gl\\) asociados a esta prueba pareada son: [[cloze_df]] (valor entero).</li>
      </ul>
  </li>

  <li><b>Valor-p y Conclusión:</b><br>
      <ul>
        <li>El valor-p arrojado por la prueba en R es: [[cloze_pval]] (a 4 decimales).</li>
        <li>A partir de los resultados obtenidos y con \\(\\alpha = [[alfa]]\\), [[cloze_decision]]. Por consiguiente, se concluye que la fuerza prensil promedio de la mano derecha [[cloze_conclusion_contexto]].</li>
      </ul>
  </li>
</ol>
'

# ==============================================================================
# 5. BUCLE GENERADOR DEL BANCO DE 15 PREGUNTAS EN R (3 Alternativas x 5)
# ==============================================================================

banco_moodle <- iniciar_quiz_xml(nombre_categoria = "CursoR/Prueba_T_Pareada_FuerzaPrensil")

opc_signos   <- c("\\neq (diferente de)", "> (mayor que)", "< (menor que)")
opc_decision <- c(
  "se rechaza la hipótesis nula (H0)",
  "no se rechaza la hipótesis nula (H0)"
)

semillas_par <- seq(901, by = 19, length.out = 15)

for (i in seq_along(semillas_par)) {
  sem <- semillas_par[i]
  set.seed(sem)
  
  # 1. Tamaño de muestra (entre 28 y 38 sujetos, similar a los 37 del ejemplo de clase)
  n <- sample(28:38, 1)
  alfa <- sample(c(0.01, 0.05), 1)
  
  # 2. Rotamos el tipo de hipótesis alternativa según la iteración:
  # Resto 1 -> two.sided (diferente/igual)
  # Resto 2 -> greater (derecha supera a izquierda)
  # Resto 0 -> less (derecha es menor que izquierda, ej: muestra de atletas zurdos)
  tipo_caso <- (i - 1) %% 3 + 1
  
  # Alternamos si se debe rechazar H0 (par/impar)
  forzar_rechazo <- (i %% 2 != 0)
  
  # 3. Simulamos datos pareados realistas (valores enteros pares/impares entre 18 y 54 kg)
  # Fuerza base de cada individuo: media ~32, DE ~9 (alta correlación entre manos)
  fuerza_base <- rnorm(n, mean = 32, sd = 8.5)
  sd_dif <- 5.0 # DE de la diferencia intra-sujeto observada en sus datos reales
  se_teorico <- sd_dif / sqrt(n)
  
  if (tipo_caso == 1) {
    # Caso Bilateral: H1: mu_D != 0
    alt_r <- "two.sided"
    pregunta_inv <- "¿existe diferencia significativa entre la fuerza prensil promedio de la mano derecha y la de la mano izquierda?"
    t_obj <- if (forzar_rechazo) sample(c(-1, 1), 1) * runif(1, 2.5, 3.6) else runif(1, -1.3, 1.3)
    opc_contexto <- c(
      "difiere significativamente de la fuerza promedio de la mano izquierda",
      "no presenta una diferencia estadísticamente significativa respecto a la de la mano izquierda"
    )
  } else if (tipo_caso == 2) {
    # Caso Unilateral Derecho: H1: mu_D > 0
    alt_r <- "greater"
    pregunta_inv <- "¿la fuerza prensil promedio de la mano derecha es significativamente mayor que la de la mano izquierda?"
    t_obj <- if (forzar_rechazo) runif(1, 2.4, 3.5) else runif(1, -0.5, 1.2)
    opc_contexto <- c(
      "es significativamente mayor que la fuerza promedio de la mano izquierda",
      "no supera de forma estadísticamente significativa a la de la mano izquierda"
    )
  } else {
    # Caso Unilateral Izquierdo: H1: mu_D < 0
    alt_r <- "less"
    pregunta_inv <- "en un grupo con predominio zurdo, ¿la fuerza prensil promedio de la mano derecha es significativamente menor que la de la mano izquierda?"
    t_obj <- if (forzar_rechazo) runif(1, -3.5, -2.4) else runif(1, -1.2, 0.5)
    opc_contexto <- c(
      "es significativamente menor que la fuerza promedio de la mano izquierda",
      "no es significativamente menor que la de la mano izquierda"
    )
  }
  
  # Generamos los vectores enteros de Mano Izquierda y Mano Derecha
  mi <- round(pmax(16, pmin(54, fuerza_base)))
  dif_sim <- round(rnorm(n, mean = t_obj * se_teorico, sd = sd_dif))
  md <- pmax(16, pmin(56, mi + dif_sim))
  
  # 4. Ejecutamos t.test(..., paired = TRUE) exactamente sobre los vectores enteros
  prueba_t <- t.test(md, mi, paired = TRUE, alternative = alt_r, conf.level = 1 - alfa)
  
  d_bar <- as.numeric(prueba_t$estimate)
  t0    <- as.numeric(prueba_t$statistic)
  gl    <- as.numeric(prueba_t$parameter)
  pval  <- as.numeric(prueba_t$p.value)
  
  # 1 = Rechaza H0 / Sí hay efecto; 2 = No rechaza H0 / No hay efecto
  idx_dec <- if (pval < alfa) 1 else 2
  
  # 5. Lista de respuestas dinámicas para inyectar en la plantilla
  respuestas_iter <- list(
    n                       = n,
    alfa                    = alfa,
    pregunta_investigacion  = pregunta_inv,
    TABLA_DATOS             = generar_tabla_pareada_html(md, mi),
    vec_md                  = paste(md, collapse = ", "),
    vec_mi                  = paste(mi, collapse = ", "),
    
    # Campos Cloze
    cloze_signo_h1          = cloze_mch(resp = tipo_caso, opciones = opc_signos, peso = 1),
    cloze_bool_paired       = cloze_sa(c("TRUE", "T"), peso = 1),
    cloze_arg_alt           = cloze_sa(c(alt_r), peso = 1),
    cloze_d_bar             = cloze_num(d_bar, digresp = 3, error = 0.01, peso = 1),
    cloze_t0                = cloze_num(t0, digresp = 3, error = 0.02, peso = 2),
    cloze_df                = cloze_num(gl, digresp = 0, error = 0, peso = 1),
    cloze_pval              = cloze_num(pval, digresp = 4, error = 0.005, peso = 2),
    cloze_decision          = cloze_mch(resp = idx_dec, opciones = opc_decision, peso = 1),
    cloze_conclusion_contexto = cloze_mch(resp = idx_dec, opciones = opc_contexto, peso = 1)
  )
  
  etiqueta_alt <- switch(alt_r, "two.sided" = "Bilateral", "greater" = "Mayor", "less" = "Menor")
  nombre_preg  <- glue("PH_Pareada_FuerzaPrensil_v{sprintf('%02d', i)}_{etiqueta_alt}_n{n}")
  
  crear_pregunta_cloze(
    moodle_quiz      = banco_moodle,
    nombre_pregunta  = nombre_preg,
    plantilla_texto  = plantilla_enunciado_pareada,
    lista_respuestas = respuestas_iter
  )
}

# 6. Exportamos el archivo XML final listo para Moodle
guardar_quiz_xml(banco_moodle, "Banco_15_Preguntas_T_Pareada_FuerzaPrensil_R.xml")
```

## Prueba de igauldad de medias con verificación de igualdad de varianzas 

Prueba de hipótesis sobre igualdad de medias con prueba de homogeneidad de variaza y desición de cuál usar.   

```r
## Ojo 

library(XML)
library(tidyverse)
library(glue)

# ==============================================================================
# 1. GENERADORES ATÓMICOS DE SINTAXIS CLOZE (Con soporte para #Feedback)
# ==============================================================================

cloze_num <- function(resp, error = 0.0001, peso = 1, digresp = 4, diger = digresp) {
  peso_int <- as.integer(peso)
  val_formated <- formatC(as.numeric(resp), digits = digresp, format = "f")
  err_formated <- formatC(as.numeric(error), digits = diger, format = "f")
  as.character(glue("{{{peso_int}:NM:={val_formated}:{err_formated}}}"))
}

cloze_mch <- function(resp = 1, opciones, feedback = NULL, peso = 1, tipo = "MULTICHOICE") {
  if (resp < 1 | resp > length(opciones)) stop("Índice de respuesta fuera de rango.")
  peso_int <- as.integer(peso)
  
  # Si no se especifica vector de feedback, asignamos #Correcto y #Incorrecto por defecto
  if (is.null(feedback)) {
    feedback <- ifelse(seq_along(opciones) == resp, "Correcto", "Incorrecto")
  }
  
  opciones_formateadas <- pmap_chr(
    list(opc = opciones, idx = seq_along(opciones), fb = feedback),
    function(opc, idx, fb) {
      fb_str <- if (!is.na(fb) && nzchar(fb)) paste0("#", fb) else ""
      if (idx != resp) paste0("~%0%", opc, fb_str) else paste0("~=", opc, fb_str)
    }
  )
  as.character(glue("{{{peso_int}:{tipo}:{paste0(opciones_formateadas, collapse = '')}}}"))
}

cloze_sa <- function(respuestas_correctas, peso = 1, case_sensitive = TRUE) {
  peso_int <- as.integer(peso)
  tipo <- if (case_sensitive) "SAC" else "SA"
  resp_str <- paste0("~=", respuestas_correctas, collapse = "")
  resp_str <- sub("^~", "", resp_str)
  as.character(glue("{{{peso_int}:{tipo}:{resp_str}}}"))
}

# Función auxiliar para clasificar el p-valor en las 5 categorías de su ejemplo
clasificar_pvalor <- function(pval) {
  if (pval < 0.001) {
    return(1) # "Casi cero (menor que 0.001)"
  } else if (pval < 0.01) {
    return(2) # "Menor que 0.01 pero mayor o igual que 0.001"
  } else if (pval < 0.05) {
    return(3) # "Menor que 0.05 pero mayor o igual que 0.01"
  } else if (pval < 0.10) {
    return(4) # "Menor que 0.10 pero mayor o igual que 0.05"
  } else {
    return(5) # "Mayor o igual que 0.10"
  }
}

# ==============================================================================
# 2. INICIALIZADOR DEL CONTENEDOR XML Y GUARDADO
# ==============================================================================

iniciar_quiz_xml <- function(nombre_categoria) {
  doc <- newXMLDoc()
  root <- newXMLNode("quiz", doc = doc)
  
  question_cat <- newXMLNode("question", attrs = list(type = "category"), parent = root)
  category <- newXMLNode("category", parent = question_cat)
  newXMLNode("text", nombre_categoria, parent = category)
  info <- newXMLNode("info", attrs = list(format = "html"), parent = question_cat)  
  newXMLNode("text", "Banco Prueba F y Prueba T (Masa Muscular) generado desde R", parent = info)
  newXMLNode("idnumber", parent = question_cat)
  
  return(list(doc = doc, root = root))
}

crear_pregunta_cloze <- function(moodle_quiz, nombre_pregunta, plantilla_texto, lista_respuestas, plantilla_tabla = NULL) {
  if (!is.null(plantilla_tabla)) {
    tabla_renderizada <- glue_data(lista_respuestas, plantilla_tabla, .open = "[[", .close = "]]")
    lista_respuestas$TABLA_EVALUACION <- as.character(tabla_renderizada)
  }
  
  texto_final_html <- glue_data(lista_respuestas, plantilla_texto, .open = "[[", .close = "]]")
  
  questionpr <- newXMLNode("question", attrs = list(type = "cloze"), parent = moodle_quiz$root)
  nombrepr <- newXMLNode("name", parent = questionpr)
  newXMLNode("text", nombre_pregunta, parent = nombrepr)
  
  questiontext <- newXMLNode("questiontext", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", as.character(texto_final_html), parent = questiontext)
  
  generalfeedback <- newXMLNode("generalfeedback", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", "", parent = generalfeedback)
  
  newXMLNode("penalty", "0.3333333", parent = questionpr)
  newXMLNode("hidden", "0", parent = questionpr)
  newXMLNode("idnumber", parent = questionpr)
}

guardar_quiz_xml <- function(moodle_quiz, archivo_salida) {
  saveXML(moodle_quiz$doc, file = archivo_salida, encoding = "UTF-8")
  cat(glue("Archivo '{archivo_salida}' generado exitosamente.\n"))
}

# ==============================================================================
# 3. GENERADOR DE TABLA HTML PARA LOS DATOS DE MASA MUSCULAR
# ==============================================================================

generar_tabla_grupos_html <- function(masc, fem) {
  str_m <- paste(formatC(masc, digits = 2, format = "f"), collapse = ", ")
  str_f <- paste(formatC(fem, digits = 2, format = "f"), collapse = ", ")
  
  glue('<style type="text/css">
.tg-masa {border-collapse:collapse; border-color:#9ABAD9; border-spacing:0; margin:12px 0; width:100%;}
.tg-masa th {background-color:#2e7d32; border:1px solid #9ABAD9; color:#fff; font-family:Arial, sans-serif; font-size:14px; padding:8px 12px; text-align:left; width:24%;}
.tg-masa td {background-color:#f1f8e9; border:1px solid #9ABAD9; color:#222; font-family:monospace; font-size:14px; padding:8px 12px; text-align:left;}
</style>
<table class="tg-masa">
  <tbody>
    <tr>
      <th>Masculino (Población 1, n = [[length(masc)]])</th>
      <td>[[str_m]]</td>
    </tr>
    <tr>
      <th>Femenino (Población 2, n = [[length(fem)]])</th>
      <td>[[str_f]]</td>
    </tr>
  </tbody>
</table>', .open = "[[", .close = "]]")
}

# ==============================================================================
# 4. PLANTILLA NARRATIVA ESTILO "bebes2.txt" ADAPTADA A FISIOLOGÍA Y R
# ==============================================================================

plantilla_enunciado_narrativo <- '
<p>En un estudio de fisiología del ejercicio se evaluó un grupo de <b>[[n_total]]</b> personas registrando, entre otras variables, el sexo del participante (<code>Masculino</code> o <code>Femenino</code>) y su <b>masa muscular</b> total en kilogramos (<code>masa_muscular</code>). Los datos obtenidos se presentan en la siguiente tabla (y en el bloque de código inferior para que pueda copiarlos directamente a <b>R</b>):</p>

[[TABLA_DATOS]]

<pre style="background-color:#f4f6f9; border-left:4px solid #2e7d32; padding:10px; font-family:monospace; font-size:13px; color:#222; overflow-x:auto;"><code>hombres &lt;- c([[vec_m]])
mujeres &lt;- c([[vec_f]])</code></pre>

<p>[[contexto_problema]] Para comprobar esta afirmación usando un nivel de significancia \\(\\alpha = [[alfa]]\\), deseamos probar la hipótesis:</p>

<p style="text-align:center;">\\(H_0: \\mu_1 = \\mu_2 \\quad \\text{versus} \\quad H_1: \\mu_1 [[simbolo_h1_tex]] \\mu_2\\)</p>

<p>donde \\(\\mu_1\\) es el promedio poblacional de masa muscular de los hombres (<code>hombres</code>, grupo 1) y \\(\\mu_2\\) es el promedio poblacional de masa muscular de las mujeres (<code>mujeres</code>, grupo 2).</p>

<hr style="border:0; border-top:1px solid #ccc; margin:15px 0;">

<p><b>1. Prueba previa sobre las varianzas:</b><br>
Antes de probar la hipótesis sobre las medias hay que decidir si se asume que las varianzas de las dos poblaciones son iguales o son distintas. Para ello, ejecutamos en R la función <code>[[cloze_func_var]](hombres, mujeres)</code> para contrastar \\(H_0: \\sigma_1^2 = \\sigma_2^2 \\text{ versus } H_1: \\sigma_1^2 \\neq \\sigma_2^2\\).</p>

<p>El valor del estadístico de prueba para la hipótesis de igualdad de varianzas es \\(F = \\) [[cloze_f0]], con [[cloze_df_num]] grados de libertad en el numerador y [[cloze_df_den]] grados de libertad en el denominador, y el p-valor de la prueba es [[cloze_pval_f]]. Por lo tanto, al nivel \\(\\alpha = [[alfa]]\\), podemos asumir que [[cloze_asume_var]].</p>

<p><b>2. Prueba sobre la igualdad de medias en R:</b><br>
De acuerdo con los resultados de la prueba anterior, llevaremos a cabo la prueba de medias asumiendo [[cloze_tipo_var]], por lo que en la función <code>t.test(hombres, mujeres, var.equal = [[cloze_bool_vareq]], alternative = "[[cloze_arg_alt]]")</code> configuramos los argumentos correspondientes.</p>

<p>El valor del estadístico de prueba resultante es \\(t = \\) [[cloze_t0]] con [[cloze_df_t]] grados de libertad y el p-valor de la prueba es [[cloze_rango_pval]], por lo tanto [[cloze_rechazo_t]].</p>

<p><b>3. Conclusión del estudio:</b><br>
De acuerdo con los resultados de la prueba podemos concluir que, en la población estudiada, [[cloze_conclusion_final]].</p>
'

# ==============================================================================
# 5. BUCLE GENERADOR DEL BANCO DE 15 PREGUNTAS EN R
# ==============================================================================

banco_moodle <- iniciar_quiz_xml(nombre_categoria = "CursoR/Banco_MasaMuscular_VarTest_TTest")

# Opciones estandarizadas inspiradas en su ejemplo de referencia
opc_asume_var <- c(
  "la varianza de la población de hombres es igual a la de la población de mujeres",
  "la varianza de la población de hombres es distinta de la de la población de mujeres"
)
opc_tipo_var <- c("Varianzas iguales", "Varianzas distintas")

opc_rangos_p <- c(
  "casi cero (menor que 0.001)",
  "menor que 0.01 pero mayor o igual que 0.001",
  "menor que 0.05 pero mayor o igual que 0.01",
  "menor que 0.10 pero mayor o igual que 0.05",
  "mayor o igual que 0.10"
)

opc_rechazo <- c("se rechaza la hipótesis nula", "no se rechaza la hipótesis nula")

opc_conclusiones <- c(
  "los hombres tienen, en promedio, mayor masa muscular que las mujeres",
  "los hombres evaluados tienen, en promedio, menor masa muscular que las mujeres deportistas",
  "existe una diferencia significativa entre la masa muscular promedio de hombres y mujeres",
  "no hay evidencia estadística suficiente para afirmar que los promedios de masa muscular difieran según lo planteado en H1"
)

semillas <- seq(1401, by = 29, length.out = 15)

for (i in seq_along(semillas)) {
  sem <- semillas[i]
  set.seed(sem)
  
  # Tamaños de muestra similares a los datos de clase (n1 ~ 20-28, n2 ~ 11-18)
  n1   <- sample(20:28, 1)
  n2   <- sample(11:18, 1)
  alfa <- 0.05
  
  # Alternamos si las varianzas son iguales (impar) o distintas (par)
  forzar_var_iguales <- (i %% 2 != 0)
  
  # Rotamos las 3 hipótesis alternativas: 1 = "!=", 2 = ">", 3 = "<"
  tipo_h1 <- (i - 1) %% 3 + 1
  
  # Alternamos si se rechaza H0 en la prueba t (2 de cada 3 rechazan, 1 de cada 3 no rechaza)
  forzar_rechazo_t <- (i %% 3 != 0)
  
  repeat {
    # Desviaciones estándar según homocedasticidad/heterocedasticidad
    if (forzar_var_iguales) {
      sd1 <- 6.2
      sd2 <- 5.8
    } else {
      sd1 <- 10.2
      sd2 <- 3.9
    }
    
    if (tipo_h1 == 1) {
      # Caso Bilateral (neq)
      simbolo_h1 <- "\\neq"
      alt_r      <- "two.sided"
      contexto   <- "Se desea investigar si factores hormonales y de entrenamiento generan una diferencia en la masa muscular promedio entre hombres y mujeres."
      mu2 <- 37.5
      mu1 <- if (forzar_rechazo_t) 48.2 else 38.8
      idx_concl_esperada <- if (forzar_rechazo_t) 3 else 4
    } else if (tipo_h1 == 2) {
      # Caso Unilateral Derecho (>)
      simbolo_h1 <- ">"
      alt_r      <- "greater"
      contexto   <- "En fisiología del ejercicio se afirma que, debido a diferencias endocrinas, los hombres presentan en promedio una masa muscular mayor que las mujeres."
      mu2 <- 37.5
      mu1 <- if (forzar_rechazo_t) 48.5 else 38.6
      idx_concl_esperada <- if (forzar_rechazo_t) 1 else 4
    } else {
      # Caso Unilateral Izquierdo (<)
      simbolo_h1 <- "<"
      alt_r      <- "less"
      contexto   <- "En un estudio especial se comparó un grupo de hombres sedentarios de oficina con un grupo de mujeres atletas de halterofilia, planteándose que en esta población los hombres tienen menor masa muscular promedio que las mujeres."
      mu2 <- 47.0
      mu1 <- if (forzar_rechazo_t) 37.5 else 46.0
      idx_concl_esperada <- if (forzar_rechazo_t) 2 else 4
    }
    
    hombres <- round(pmax(22, pmin(72, rnorm(n1, mean = mu1, sd = sd1))), 2)
    mujeres <- round(pmax(25, pmin(62, rnorm(n2, mean = mu2, sd = sd2))), 2)
    
    # Verificamos con var.test y t.test que los p-valores no queden en zonas grises cercanas a 0.05
    vt_chk <- var.test(hombres, mujeres)
    cond_var_ok <- if (forzar_var_iguales) (vt_chk$p.value > 0.15) else (vt_chk$p.value < 0.015)
    
    tt_chk <- t.test(hombres, mujeres, var.equal = forzar_var_iguales, alternative = alt_r)
    cond_t_ok <- if (forzar_rechazo_t) (tt_chk$p.value < 0.02) else (tt_chk$p.value > 0.15)
    
    if (cond_var_ok && cond_t_ok) break
  }
  
  # --- Cálculos exactos sobre los vectores redondeados ---
  res_f   <- var.test(hombres, mujeres)
  f_stat  <- as.numeric(res_f$statistic)
  df_num  <- as.numeric(res_f$parameter[1])
  df_den  <- as.numeric(res_f$parameter[2])
  pval_f  <- as.numeric(res_f$p.value)
  
  var_iguales_bool <- (pval_f >= alfa)
  idx_var          <- if (var_iguales_bool) 1 else 2
  resp_vareq_str   <- if (var_iguales_bool) c("TRUE", "T") else c("FALSE", "F")
  
  res_t   <- t.test(hombres, mujeres, var.equal = var_iguales_bool, alternative = alt_r)
  t_stat  <- as.numeric(res_t$statistic)
  df_t    <- as.numeric(res_t$parameter)
  pval_t  <- as.numeric(res_t$p.value)
  
  idx_rango_p <- clasificar_pvalor(pval_t)
  idx_rechazo <- if (pval_t < alfa) 1 else 2
  idx_concl   <- if (pval_t < alfa) idx_concl_esperada else 4
  
  # Si var.equal = TRUE, df_t es entero; si es FALSE (Welch), tiene decimales
  dig_dft <- if (var_iguales_bool) 0 else 2
  err_dft <- if (var_iguales_bool) 0.5 else 0.05
  
  respuestas_iter <- list(
    n_total           = n1 + n2,
    alfa              = alfa,
    contexto_problema = contexto,
    simbolo_h1_tex    = simbolo_h1,
    TABLA_DATOS       = generar_tabla_grupos_html(hombres, mujeres),
    vec_m             = paste(formatC(hombres, digits = 2, format = "f"), collapse = ", "),
    vec_f             = paste(formatC(mujeres, digits = 2, format = "f"), collapse = ", "),
    
    # Parte 1: Prueba F de varianzas (var.test)
    cloze_func_var    = cloze_sa(c("var.test", "stats::var.test"), peso = 1),
    cloze_f0          = cloze_num(f_stat, digresp = 4, error = 0.01, peso = 1),
    cloze_df_num      = cloze_num(df_num, digresp = 0, error = 0, peso = 1),
    cloze_df_den      = cloze_num(df_den, digresp = 0, error = 0, peso = 1),
    cloze_pval_f      = cloze_num(pval_f, digresp = 4, error = 0.01, peso = 1),
    cloze_asume_var   = cloze_mch(resp = idx_var, opciones = opc_asume_var, peso = 2),
    
    # Parte 2: Decisión sobre var.equal y Prueba T (t.test)
    cloze_tipo_var    = cloze_mch(resp = idx_var, opciones = opc_tipo_var, peso = 2),
    cloze_bool_vareq  = cloze_sa(resp_vareq_str, peso = 1),
    cloze_arg_alt     = cloze_sa(c(alt_r), peso = 1),
    cloze_t0          = cloze_num(t_stat, digresp = 4, error = 0.05, peso = 2),
    cloze_df_t        = cloze_num(df_t, digresp = dig_dft, error = err_dft, peso = 1),
    cloze_rango_pval  = cloze_mch(resp = idx_rango_p, opciones = opc_rangos_p, peso = 2),
    cloze_rechazo_t   = cloze_mch(resp = idx_rechazo, opciones = opc_rechazo, peso = 2),
    
    # Parte 3: Conclusión contextualizada
    cloze_conclusion_final = cloze_mch(resp = idx_concl, opciones = opc_conclusiones, peso = 3)
  )
  
  etq_var <- if (var_iguales_bool) "VarIguales" else "VarDistintas"
  etq_alt <- switch(alt_r, "two.sided" = "Bilateral", "greater" = "Mayor", "less" = "Menor")
  nombre_preg <- glue("PH_Narrativa_MasaMuscular_v{sprintf('%02d', i)}_{etq_var}_{etq_alt}")
  
  crear_pregunta_cloze(
    moodle_quiz      = banco_moodle,
    nombre_pregunta  = nombre_preg,
    plantilla_texto  = plantilla_enunciado_narrativo,
    lista_respuestas = respuestas_iter
  )
}

# 6. Guardamos el archivo XML listo para Moodle
guardar_quiz_xml(banco_moodle, "Banco_15_Preguntas_Narrativas_MasaMuscular_R.xml")
```
