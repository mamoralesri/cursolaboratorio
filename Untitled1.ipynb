```r
library(XML)
library(tidyverse)
library(readxl)
library(glue)

# ==============================================================================
# FASE 1: FUNCIONES ATÓMICAS REFACTORIZADAS
# ==============================================================================

# Genera respuestas numéricas Cloze
pregnum <- function(resp, error = 0.0001, peso = 1, digresp = 4, diger = digresp) {
  val_formated <- formatC(resp, digits = digresp, format = "f")
  err_formated <- formatC(error, digits = diger, format = "f")
  glue("{{1:NM:={val_formated}:{err_formated}}}")
}

# Genera respuestas de opción múltiple Cloze (Protege espacios internos del código)
premch <- function(resp = 1, opciones, peso = 1) {
  if (resp < 1 | resp > length(opciones)) {
    stop("La respuesta debe ser un número entre 1 y el número de opciones")
  }
  
  opciones_formateadas <- map2_chr(opciones, seq_along(opciones), function(opc, idx) {
    if (idx != resp) paste0("~%0%", opc) else paste0("~=", opc)
  })
  
  # Concatenamos de forma segura sin destruir espacios intermedios del código de R
  glue("{{1:MULTICHOICE:{paste0(opciones_formateadas, collapse = '')}}}")
}

# ==============================================================================
# FASE 2: INICIALIZACIÓN DEL ÁRBOL XML DE MOODLE
# ==============================================================================
NombreCat <- "Limpieza/parse_num"
doc <- newXMLDoc()
root <- newXMLNode("quiz", doc = doc)

# Nodo de categoría obligatorio
question_cat <- newXMLNode("question", attrs = list(type = "category"), parent = root)
category <- newXMLNode("category", parent = question_cat)
newXMLNode("text", NombreCat, parent = category)
info <- newXMLNode("info", attrs = list(format = "html"), parent = question_cat)  
newXMLNode("text", "texto", parent = info)
newXMLNode("idnumber", parent = question_cat)

# Función para inyectar nodos Cloze al árbol global
addquestion <- function(NomPr, textopregunta) {
  questionpr <- newXMLNode("question", attrs = list(type = "cloze"), parent = root)
  nombrepr <- newXMLNode("name", parent = questionpr)
  newXMLNode("text", NomPr, parent = nombrepr)
  questiontext <- newXMLNode("questiontext", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", textopregunta, parent = questiontext)
  generalfeedback <- newXMLNode("generalfeedback", attrs = list(format = "html"), parent = questionpr)
  newXMLNode("text", parent = generalfeedback)
  newXMLNode("penalty", "0.3333333", parent = questionpr)
  newXMLNode("hidden", "0", parent = questionpr)
  newXMLNode("idnumber", parent = questionpr)
}

# ==============================================================================
# FASE 3: EL MOTOR AUTOMÁTICO DE PREGUNTAS (LA MAGIA)
# ==============================================================================

generar_pregunta_columna <- function(data, columna, id_semilla, opc_dplyr, opc_stats, idx_mch) {
  
  # 1. Cálculos de control y métricas descriptivas bajo demanda
  serie_datos <- data[[columna]]
  
  nas_count <- sum(is.na(serie_datos))
  n_val <- length(serie_datos)
  min_val <- min(serie_datos, na.rm = TRUE)
  q1_val <- quantile(serie_datos, probs = 0.25, na.rm = TRUE)
  med_val <- median(serie_datos, na.rm = TRUE)
  med_arn <- mean(serie_datos, na.rm = TRUE)
  
  # 2. Construcción sintáctica de las respuestas Cloze de Moodle
  cloze_nas   <- pregnum(nas_count, digresp = 0, error = 0)
  cloze_n     <- pregnum(n_val, digresp = 0, error = 0)
  cloze_min   <- pregnum(min_val, digresp = 0, error = 0)
  cloze_q1    <- pregnum(q1_val, digresp = 2, error = 0.01)
  cloze_med   <- pregnum(med_val, digresp = 2, error = 0.01)
  cloze_media <- pregnum(med_arn, digresp = 3, error = 0.005)
  
  # 3. Construcción del HTML dinámico usando bloques estructurados con 'glue'
  tabla_html <- glue('
    <table style="border-collapse:collapse;border-color:#aaa;border-spacing:0" class="tg">
      <thead>
        <tr>
          <th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;padding:10px 5px;text-align:left;">N</th>
          <th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;padding:10px 5px;text-align:left;">Min</th>
          <th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;padding:10px 5px;text-align:left;">Q1</th>
          <th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;padding:10px 5px;text-align:left;">Mediana</th>
          <th style="background-color:#f38630;border-color:inherit;border-style:solid;border-width:1px;color:#fff;font-family:Arial, sans-serif;font-size:14px;font-weight:normal;padding:10px 5px;text-align:left;">Media</th>
        </tr>
      </thead>
      <tbody>
        <tr>
          <td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;padding:10px 5px;">{cloze_n}</td>
          <td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;padding:10px 5px;">{cloze_min}</td>
          <td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;padding:10px 5px;">{cloze_q1}</td>
          <td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;padding:10px 5px;">{cloze_med}</td>
          <td style="background-color:#fff;border-color:inherit;border-style:solid;border-width:1px;color:#333;font-family:Arial, sans-serif;font-size:14px;padding:10px 5px;">{cloze_media}</td>
        </tr>
      </tbody>
    </table>')

  # Ensamble estructural del texto completo del ítem
  texto_completo <- glue("
    El archivo <code>DatosEstudiantes.xlsx</code> contiene datos tomados de estudiantes de primer semestre durante un experimento que buscaba evaluar el nivel de satisfacción de los estudiantes con el servicio prestado por la Universidad. Haga la lectura de los datos y responda:<br><br>
    Cuántos datos faltantes hay en la columna <code>{columna}</code>: {cloze_nas}<br><br>
    Un posible código, en lenguaje <code>R</code>, usando la librería <code>tidyverse</code> y el <code>pipe</code> para obtener la respuesta a la pregunta anterior es:<br>
    <pre>library(tidyverse)\nlibrary(readxl)\ndf = read_excel('DatosEstudiantes.xlsx')\ndf = {premch(2, opc_dplyr)}({columna} = {premch(idx_mch, opc_dplyr)}({columna}) |> {premch(7, opc_dplyr)}(sum(is.na({columna}))))</pre><br>
    Complete la siguiente tabla con estadísticas descriptivas básicas de la columna <code>{columna}</code>:<br>
    {tabla_html}<br>
    Un posible código, en lenguaje <code>R</code>, usando la librería <code>tidyverse</code> y el <code>pipe</code> para obtener la respuesta a la tabla anterior es:<br>
    <pre>library(tidyverse)\nlibrary(readxl)\ndf = read_excel('DatosEstudiantes.xlsx')\ndf |> {premch(7, opc_dplyr)}(N = n(), Min = {premch(1, opc_stats)}({columna}, na.rm = TRUE), Q1 = {premch(4, opc_stats)}({columna}, na.rm = TRUE), {premch(9, opc_stats)} = {premch(3, opc_stats)}({columna}, na.rm = TRUE), {premch(8, opc_stats)} = {premch(2, opc_stats)}({columna}, na.rm = TRUE))</pre>
  ")

  # 4. Empujar el ítem al XML
  nombre_final_pregunta <- paste("Filter", id_semilla, sep = " ")
  addquestion(NomPr = nombre_final_pregunta, textopregunta = texto_completo)
}

# Definición de bancos de distractores comunes (opciones de selección múltiple)
opc0 <- c('is_na(Peso)','mutate','n()','mean(Peso)','parse_number','Peso','summarise','min','Nhermanos','Edad') 
opc1 <- c('min','mean','median','quantile','max','var','sd','Media','Mediana')

# Lectura y limpieza rápida inicial (Su lógica de negocio)
df <- read_excel("~/Documents/Cursos/Laboratorio/DatosEstudiantes.xlsx")

df_limpio <- df %>%
  mutate(
    Peso = parse_number(Peso),
    Estatura = parse_number(Estatura),
    Estatura = if_else(Estatura > 2.0, Estatura / 100, Estatura)
  )

# ==============================================================================
# EJECUCIÓN: CREAR PREGUNTAS EN UNA SOLA LÍNEA DE CÓDIGO
# ==============================================================================

# Pregunta 1: Para la columna 'Peso' (El distractor de parse_number está en el índice 5 de opc0)
generar_pregunta_columna(df_limpio, columna = "Peso", id_semilla = 1, opc_dplyr = opc0, opc_stats = opc1, idx_mch = 5)

# Pregunta 2: Para la columna 'Nhermanos' (El distractor de la variable es 'Nhermanos', índice 9 de opc0)
generar_pregunta_columna(df_limpio, columna = "Nhermanos", id_semilla = 2, opc_dplyr = opc0, opc_stats = opc1, idx_mch = 9)

# Pregunta 3: ¿Quiere agregar la columna 'Estatura' mañana? ¡Es así de fácil!
# generar_pregunta_columna(df_limpio, columna = "Estatura", id_semilla = 3, opc_dplyr = opc0, opc_stats = opc1, idx_mch = 6)

# ==============================================================================
# GUARDAR EL ARCHIVO GENERAL FINAL
# ==============================================================================
saveXML(doc, file = "PreguntasParse.xml") 
```
