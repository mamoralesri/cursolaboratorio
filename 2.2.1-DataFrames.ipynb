## Contenedores de datos en python

Pyton ofrece varias estructuras para el almacenamiento de datos, entre ellas tenemos

### DataFrames

*pandas* es el paquete de python que proporciona estructuras de datos útiles para análisis de estadístico, ademas de funciones que facilitan la entrada, organización y manipulación de datos. Un `DataFrame` es una estructura de datos bidimensional con columnas de  (potencialmente) distintos tipos de datos. Los `DataFrame` son los objetos más comunmente usados en *pandas*. 



#### Creación y operaciones con data-frames

Es común importar el paquete *pandas* como `pd` mediante la línea `import pandas as pd`. Eso es lo primero que hacemos en las siguientes lineas de código, luego creamos un `DataFrame` de nombre `df` con dos columnas y cuatro filas, nótese que lo hacemos a partir de un diccionario.  

```python
import pandas as pd
df = pd.DataFrame({"col1":[1,3,11,2], "col2":[9,23,0,2]}) 
print(type(df))
df
```

### Selección de filas y columnas con `iloc`

El método `iloc` selecciona tanto filas como columnas del `DataFrame` basado en la posición del elemento (fila o columna). La sintaxis es de la forma `df.iloc[<fila>,<columna>]` donde `df` es un `DataFrame`, `fila` puede ser una lista con enteros indicando la posición (índice) de las filas a seleccionar o un número entero, de manera similar para `columna`. Cuando se entrega un solo valor, por ejemplo `df.iloc[valor]` la selección se hace sobre las filas, veamos con ejemplos su funcionamiento. 

```python
# selecciona la primera fila del data
#print(df.iloc[0])  
# equivalente a df.iloc[0,]
#df.iloc[0,] 
# selecciona la segunda fila del data
#print(df.iloc[1]) 
# equivalente a df.iloc[1,]
# selecciona la última fila del data
#print(df.iloc[-1]) 
# las primeras 3 filas 
df.iloc[0:3]
```

```python
# selecciona la primera columna del data con todas las filas 
print(df.iloc[:,0])
#equivalente a  df.iloc[0:5,0]
# selecciona la tercera (indice 2) columna del data con todas las filas 
print(df.iloc[:,2])
# selecciona la última columna del data con todas las filas 
df.iloc[:,-1] 
```

```python
# Las dos primeras columnas con todas las filas 
df.iloc[:,0:2]
```

```python
df.iloc[[0,2,4], [0,2]]  
```

### Selección de filas y columnas con `loc` 

Con el método `loc`se puede acceder a un grupo de filas o columnas de la tabla de datos mediante etiquetas o un arreglo cuyos elementos son valores lógicos.  

En el siguiente ejemplo se seleccionan aquellas filas para las cuales se verifica que los valores de `col1` son mayores que 4

```python
df.loc[df.Edad>=18] 
```

```python
df.loc[lambda df: df['Genero']=="M"] 
```

```python
df.loc[ (df.Edad>=17) & (df.Genero=="M")]  
```

```python
def pos_percentil(datos,k):
    n=datos.size 
    if(n*k==int(n*k)):
        Pe=(np.sort(datos)[int(n*k)] + np.sort(datos)[int(n*k)+1] )/2
        return("\\( n \\times k ="+str(int(n*k))+ "\\), que es un número entero, luego debemos promediar los datos en las posiciones "+str(int(n*k))+" y "+str(int(n*k)+1)+", es decir, promediar "
                    +str(np.sort(datos)[int(n*k)])+" y "+str(np.sort(datos)[int(n*k)+1])+", así las cosas, el percentil buscado es \\(\\frac{"+str(np.sort(datos)[int(n*k)])+"+"+str(np.sort(datos)[int(n*k)+1])+"}{2}="+str(Pe)+"\\)" )     
    else:
        Pe=np.sort(datos)[int(np.ceil(n*k) )]
        return("\\( n \\times k = " +str(n*k)+"\\), que no es un número entero, luego el percentil requerido es el dato que está en la posición "+str(int( np.ceil(n*k)))+", asi las cosas, el percentil buscado es "+str(Pe) )
```
